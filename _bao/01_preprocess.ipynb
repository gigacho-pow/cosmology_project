```python
import numpy as np
import pandas as pd

bao = pd.read_csv(
    "../data/raw/bao/desi_gaussian_bao_ALL_GCcomb_mean.txt",
    sep=r"\s+",
    comment="#",
    names=["z", "value", "quantity"],
)

cov = np.loadtxt(
    "../data/raw/bao/desi_gaussian_bao_ALL_GCcomb_cov.txt"
)

bao
```

```python
assert cov.shape == (len(bao), len(bao))
assert np.isfinite(bao[["z", "value"]].values).all()
assert np.isfinite(cov).all()
assert (bao["z"] > 0).all()
assert bao["quantity"].isin(
    ["DM_over_rs", "DH_over_rs", "DV_over_rs"]
).all()
assert np.allclose(cov, cov.T)

cov = (cov + cov.T) / 2

bao.to_csv("processed/bao.csv", index=False)
np.savez_compressed("processed/bao_covariance.npz", covariance=cov)
```
